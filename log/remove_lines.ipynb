```python
def remove_lines(filepath, target_string):
    with open(filepath, 'r', encoding='utf-8') as file:
        lines = file.readlines()
    
    target_lower = target_string.lower()
    
    with open(filepath, 'w', encoding='utf-8') as file:
        for line in lines:
            if target_lower not in line.lower():
                file.write(line)

remove_lines('./@access_202608.log', '240a:61:4225:a22b:d995:632:7af2:a760')
```
