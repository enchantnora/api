```python
import os
import re

def remove_lines(filepath=None, target_string=None):
    if target_string is None:
        target_string = filepath
        filepath = None

    if not filepath:
        pattern = re.compile(r'^access_\d{4}(0[1-9]|1[0-2])\.log$')
        files = [
            f'./{f}' for f in os.listdir('.')
            if os.path.isfile(f) and pattern.match(f)
        ]
        if not files:
            print('ファイルが見つかりません')
            return
        filepath = max(files)

    with open(filepath, 'r', encoding='utf-8') as file:
        lines = file.readlines()
    
    target_lower = target_string.lower()
    
    with open(filepath, 'w', encoding='utf-8') as file:
        for line in lines:
            if target_lower not in line.lower():
                file.write(line)

remove_lines('', '240a:61:3221:35e2:4053:7a:c346:ec0')
```
